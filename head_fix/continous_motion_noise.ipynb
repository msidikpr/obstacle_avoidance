Create motion vector field 

```python

def create_motion_vector_field(size, max_norm,min_norm):
    """
    Create a 2D array of motion vectors with random norms and phase angles.

    Parameters:
        size (tuple): Size of the 2D array (rows, columns).
        max_norm (float): Maximum value for the norm of the motion vectors.

    Returns:
        norm_array (ndarray): 2D array of norms.
        angle_array (ndarray): 2D array of phase angles (in radians).
    """
    # Randomly sample norms from a uniform distribution [0, max_norm)
    norm_array = np.random.uniform(min_norm, max_norm, size)

    # Randomly sample phase angles from a uniform distribution [0, 2π)
    angle_array = np.random.uniform(0, 360, size)

    return norm_array, angle_array

# Example usage
```

```python
size = (16, 38,2)  # Size of the 2D array
# frames , rows , columns
max_norm = 1 
min_norm = 1  # Maximum norm value
norm_array, angle_array = create_motion_vector_field(size, max_norm,min_norm)
```
